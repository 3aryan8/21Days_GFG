```python
## Solved for Country
print(netflix_df['country'].isna().sum())
print(netflix_df["country"].value_counts())
```

```python
netflix_df['country'].fillna(netflix_df['country'].mode()[0] , inplace=True)
```

```python
netflix_df.isna().sum(axis=0)
```

```python
netflix_df.dropna(subset=['date_added' , 'rating'] , inplace=True)
```

```python
## Moving on to transformations. 

print(netflix_df.info())
netflix_df.head()
```

```python
# transforming date_added to date and time type

netflix_df['date_added'] = pd.to_datetime(netflix_df["date_added"], format='mixed' , dayfirst=False)
type(netflix_df['date_added'].dtype)
# Conversion sucessfull
```

```python
## For doing year wise , and monthwise analysis
## we will create new columns, 

netflix_df["year_added"] = netflix_df["date_added"].dt.year
netflix_df["month_added"] = netflix_df["date_added"].dt.month
```
